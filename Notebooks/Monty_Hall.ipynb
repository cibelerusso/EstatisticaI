Em um programa de TV, são mostradas três portas, 1, 2 e 3, fechadas. Um prêmio está escondido em uma delas. O apresentador do programa sabe qual a porta que esconde o prêmio. Um participante escolhe uma porta, e ela não é aberta. O apresentador então escolhe uma das outras portas, que não guarda o prêmio, e abre-a. O participante deve trocar de porta ou manter a sua escolha inicial?


Considere para os cálculos os eventos:

$X_i$: "A porta i guarda o prêmio".

$Y_i$: "O apresentador informa que a porta j não guarda o prêmio".


```python
import random

def jogar_monty_hall(trocar=True):
    """
    Simula uma rodada do problema de Monty Hall.

    Parâmetro:
        trocar: True  -> o participante troca de porta
                False -> o participante mantém a escolha inicial

    Retorna:
        True se o participante ganhar e False caso contrário.
    """

    portas = [1, 2, 3]

    # O prêmio é colocado aleatoriamente em uma porta
    porta_premio = random.choice(portas)

    # Escolha inicial do participante
    escolha_inicial = random.choice(portas)

    # O apresentador pode abrir uma porta que:
    # 1. não foi escolhida pelo participante;
    # 2. não contém o prêmio.
    portas_que_podem_ser_abertas = [
        porta for porta in portas
        if porta != escolha_inicial and porta != porta_premio
    ]

    porta_aberta = random.choice(portas_que_podem_ser_abertas)

    # A única porta que não foi escolhida nem aberta
    porta_restante = next(
        porta for porta in portas
        if porta != escolha_inicial and porta != porta_aberta
    )

    # O participante mantém ou troca sua escolha
    if trocar:
        escolha_final = porta_restante
    else:
        escolha_final = escolha_inicial

    # Verifica se ganhou
    return escolha_final == porta_premio


def simular(numero_de_rodadas=100):
    """
    Compara as estratégias de manter e trocar de porta.
    """

    vitorias_mantendo = 0
    vitorias_trocando = 0

    for _ in range(numero_de_rodadas):
        if jogar_monty_hall(trocar=False):
            vitorias_mantendo += 1

        if jogar_monty_hall(trocar=True):
            vitorias_trocando += 1

    probabilidade_mantendo = vitorias_mantendo / numero_de_rodadas
    probabilidade_trocando = vitorias_trocando / numero_de_rodadas

    print(f"Número de rodadas: {numero_de_rodadas:,}")
    print()
    print("Estratégia de manter:")
    print(f"  Vitórias: {vitorias_mantendo:,}")
    print(f"  Probabilidade experimental: "
          f"{probabilidade_mantendo:.4f} "
          f"({probabilidade_mantendo:.2%})")
    print()
    print("Estratégia de trocar:")
    print(f"  Vitórias: {vitorias_trocando:,}")
    print(f"  Probabilidade experimental: "
          f"{probabilidade_trocando:.4f} "
          f"({probabilidade_trocando:.2%})")


simular(1000000)
```
